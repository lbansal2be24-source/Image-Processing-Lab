# Assignment 3: Border Padding and Image Complement

```python
import cv2
import matplotlib.pyplot as plt

# Read grayscale image
gray_image = cv2.imread('grayscale.jpg', cv2.IMREAD_GRAYSCALE)

# Convert grayscale image to binary image
_, binary_image = cv2.threshold(gray_image, 127, 255, cv2.THRESH_BINARY)

# Display both images
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(gray_image, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(binary_image, cmap='gray')
plt.title('Binary Image')
plt.axis('off')

plt.show()
```

## Question 1: Border / Padding

Draw a border around the input image by padding the image with a user-defined width and color.
The operation is performed for both grayscale and binary images.

```python
# User input for border width
border_width = int(input("Enter border width in pixels: "))

# User input for border color
border_color = int(input("Enter border color (0-255): "))

# Add border to grayscale image
gray_border = cv2.copyMakeBorder(
    gray_image,
    border_width,
    border_width,
    border_width,
    border_width,
    cv2.BORDER_CONSTANT,
    value=border_color
)

# Add border to binary image
binary_border = cv2.copyMakeBorder(
    binary_image,
    border_width,
    border_width,
    border_width,
    border_width,
    cv2.BORDER_CONSTANT,
    value=border_color
)

# Display results
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(gray_border, cmap='gray')
plt.title('Grayscale Image with Border')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(binary_border, cmap='gray')
plt.title('Binary Image with Border')
plt.axis('off')

plt.show()
```

## Question 2: Image Complement

Find the complement of both the grayscale and binary images.

```python
# Find complement of grayscale image
gray_complement = 255 - gray_image

# Find complement of binary image
binary_complement = 255 - binary_image

# Display complement images
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(gray_complement, cmap='gray')
plt.title('Complement of Grayscale Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(binary_complement, cmap='gray')
plt.title('Complement of Binary Image')
plt.axis('off')

plt.show()
```

The input images were successfully processed using image processing techniques.

A user-defined border was added around both grayscale and binary images.
The border width and color were taken as input from the user.
The complement of the grayscale image was obtained by subtracting pixel values from 255.
The complement of the binary image was obtained by converting black pixels to white and white pixels to black.
Thus, both border/padding and image complement operations were successfully performed.
