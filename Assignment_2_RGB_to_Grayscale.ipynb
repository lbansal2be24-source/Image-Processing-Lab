## RGB (Color) Image to Grayscale Image

### Objective
To convert an RGB (Color) image into a grayscale image using:
1. Mean average of the three color planes (R, G, B)
2. User-defined weights for R, G, B planes

```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread('images.jpeg')

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image_rgb)
plt.title('Original RGB Image')
plt.axis('off')
plt.show()
```

## Case 1: Mean Average of Three Planes

In this case, the grayscale image is obtained by taking the mean average of the Red (R), Green (G), and Blue (B) planes.

Formula:

Grayscale = (R + G + B) / 3

```python
# Case 1: Mean average of R, G, B planes

b, g, r = cv2.split(image)

gray_mean = (r.astype(float) + g.astype(float) + b.astype(float)) / 3

gray_mean = gray_mean.astype('uint8')

plt.imshow(gray_mean, cmap='gray')
plt.title('Grayscale Image - Mean Average')
plt.axis('off')
plt.show()
```

## Case 2: User-Defined Weights

In this case, weights for the Red (R), Green (G), and Blue (B) planes are taken from the user.

Each weight must be between 0 and 1, and the sum of all three weights must be equal to 1.

Example:
R = 0.7, G = 0.2, B = 0.1

Formula:

Grayscale = R × Weight(R) + G × Weight(G) + B × Weight(B)

```python
# Case 2: User-defined weights

wr = float(input("Enter weight for Red (0-1): "))
wg = float(input("Enter weight for Green (0-1): "))
wb = float(input("Enter weight for Blue (0-1): "))

if abs(wr + wg + wb - 1) > 0.0001:
    print("Error: Sum of weights must be equal to 1.")
else:
    gray_weighted = (
        wr * r.astype(float) +
        wg * g.astype(float) +
        wb * b.astype(float)
    )

    gray_weighted = gray_weighted.astype('uint8')

    plt.imshow(gray_weighted, cmap='gray')
    plt.title('Grayscale Image - Weighted Average')
    plt.axis('off')
    plt.show()
```

## Result

The RGB image was successfully converted into grayscale images using both:
- Mean average method
- User-defined weighted average method
