## Grayscale Image to Binary Image

```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread('grayscale.jpg', cv2.IMREAD_GRAYSCALE)

plt.imshow(image, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')
plt.show()
```

## Case 1: Mean Intensity as Threshold


```python
mean_intensity = image.mean()

print("Mean Intensity =", mean_intensity)
```

```python
_, binary_image = cv2.threshold(
    image,
    mean_intensity,
    255,
    cv2.THRESH_BINARY
)

plt.imshow(binary_image, cmap='gray')
plt.title('Binary Image')
plt.axis('off')
plt.show()
```

## Case 2: User Defined Threshold

```python
threshold_value = int(input("Enter threshold value (0-255): "))

print("Threshold value =", threshold_value)
```

```python
_, binary_image_user = cv2.threshold(
    image,
    threshold_value,
    255,
    cv2.THRESH_BINARY
)

plt.imshow(binary_image_user, cmap='gray')
plt.title('Binary Image - User Defined Threshold')
plt.axis('off')
plt.show()
```
