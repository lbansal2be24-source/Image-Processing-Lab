## Question 1: Log Transformation

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
image = cv2.imread('grayscale.jpg', cv2.IMREAD_GRAYSCALE)

# Log transformation: s = c * log(1 + r), c = 1
log_image = np.log1p(image.astype(np.float32))

# Normalize to 0-255 for display
log_image = cv2.normalize(log_image, None, 0, 255, cv2.NORM_MINMAX)
log_image = log_image.astype(np.uint8)

# Display original and log transformed image
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Grayscale Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(log_image, cmap='gray')
plt.title('Log Transformed Image')
plt.axis('off')

plt.show()
```

### Result

The input grayscale image was successfully enhanced using Log Transformation with c = 1. The log transformation improved the visibility of details in darker regions of the image.

## Question 2: Power Law / Gamma Transformation

Enhance the input grayscale image using Power Law (Gamma) Transformation with c = 1.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
image = cv2.imread('grayscale.jpg', cv2.IMREAD_GRAYSCALE)

# Take gamma value from user
gamma = float(input("Enter gamma value: "))

# Power law transformation: s = c * r^gamma, c = 1
normalized = image.astype(np.float32) / 255.0
gamma_image = np.power(normalized, gamma)

# Convert back to 0-255
gamma_image = (gamma_image * 255).astype(np.uint8)

# Display original and gamma transformed image
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Grayscale Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(gamma_image, cmap='gray')
plt.title('Gamma Transformed Image')
plt.axis('off')

plt.show()
```

### Result

The input grayscale image was successfully enhanced using Power Law (Gamma) Transformation with c = 1 and gamma = 1.5. The transformation enhanced the darker regions of the image.

## Question 3: Intensity Level Slicing (Contrast Stretching)

Enhance the input image using Intensity Level Slicing (Contrast Stretching). The input intensity range and desired output intensity range are taken from the user.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
image = cv2.imread('grayscale.jpg', cv2.IMREAD_GRAYSCALE)

# Take input and output intensity ranges from user
r1 = int(input("Enter lower input intensity value: "))
r2 = int(input("Enter upper input intensity value: "))
s1 = int(input("Enter lower output intensity value: "))
s2 = int(input("Enter upper output intensity value: "))

# Contrast stretching
output = np.zeros_like(image, dtype=np.uint8)

mask = (image >= r1) & (image <= r2)

output[image < r1] = 0
output[image > r2] = 255
output[mask] = ((image[mask] - r1) * (s2 - s1) / (r2 - r1) + s1).astype(np.uint8)

# Display original and enhanced image
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Grayscale Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(output, cmap='gray')
plt.title('Intensity Level Slicing')
plt.axis('off')

plt.show()
```

### Result

The input grayscale image was successfully enhanced using intensity level slicing (contrast stretching). The selected input intensity range was mapped to the desired output intensity range, improving the contrast of the image.
