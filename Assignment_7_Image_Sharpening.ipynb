## Aim:
To sharpen the input image using:

1. Laplacian Filter
2. Sobel Gradient Operator
3. Canny Edge Detector
4. Unsharp Masking

Libraries Used: OpenCV, NumPy and Matplotlib.



```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('/content/newimage.jpeg')

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 5))
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
laplacian = cv2.Laplacian(img, cv2.CV_64F)

sharpened = img.astype(np.float64) - laplacian

sharpened = np.clip(sharpened, 0, 255).astype(np.uint8)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(sharpened, cv2.COLOR_BGR2RGB))
plt.title("Laplacian Sharpened Image")
plt.axis("off")

plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)

sobel_x = cv2.convertScaleAbs(sobel_x)
sobel_y = cv2.convertScaleAbs(sobel_y)

sobel_combined = cv2.addWeighted(sobel_x, 0.5, sobel_y, 0.5, 0)

plt.figure(figsize=(12, 5))

plt.subplot(1, 3, 1)
plt.imshow(sobel_x, cmap='gray')
plt.title("Sobel X")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(sobel_y, cmap='gray')
plt.title("Sobel Y")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(sobel_combined, cmap='gray')
plt.title("Sobel Combined")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
canny = cv2.Canny(gray, 100, 200)

plt.figure(figsize=(6, 5))
plt.imshow(canny, cmap='gray')
plt.title("Canny Edge Detection")
plt.axis("off")
plt.show()
```

```python
blurred = cv2.GaussianBlur(img, (5, 5), 3)

unsharp = cv2.addWeighted(img, 1.5, blurred, -0.5, 0)

plt.figure(figsize=(12, 5))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(blurred, cv2.COLOR_BGR2RGB))
plt.title("Gaussian Blurred Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(unsharp, cv2.COLOR_BGR2RGB))
plt.title("Unsharp Masking")
plt.axis("off")

plt.tight_layout()
plt.show()
```

## Result:

The input image was successfully sharpened using:

1. Laplacian Filter
2. Sobel Gradient Operator
3. Canny Edge Detector
4. Unsharp Masking

Different edge detection and sharpening techniques were applied to observe the improvement in image details and edges.
