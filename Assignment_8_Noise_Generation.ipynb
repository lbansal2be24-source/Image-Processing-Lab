## Aim:
To generate different types of noise using their probability distribution equations and compare them with built-in functions.

1. Uniform Noise
2. Gaussian Noise
3. Erlang Noise
4. Exponential Noise
5. Rayleigh Noise

Libraries Used: OpenCV, NumPy and Matplotlib.


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('/content/newimage.jpeg')

img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 5))
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
a = -30
b = 30

u = np.random.rand(*img.shape)

uniform_manual = a + (b - a) * u

uniform_builtin = np.random.uniform(a, b, img.shape)

noisy_manual = np.clip(img.astype(float) + uniform_manual, 0, 255).astype(np.uint8)

noisy_builtin = np.clip(img.astype(float) + uniform_builtin, 0, 255).astype(np.uint8)

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(noisy_manual)
plt.title("Uniform Noise - Equation")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(noisy_builtin)
plt.title("Uniform Noise - Built-in")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.hist(uniform_manual.ravel(), bins=50)
plt.title("Manual Uniform Noise Histogram")

plt.subplot(2, 2, 4)
plt.hist(uniform_builtin.ravel(), bins=50)
plt.title("Built-in Uniform Noise Histogram")

plt.tight_layout()
plt.show()
```

```python
mean = 0
sigma = 20

u1 = np.random.rand(*img.shape)
u2 = np.random.rand(*img.shape)

gaussian_manual = sigma * np.sqrt(-2 * np.log(u1)) * np.cos(2 * np.pi * u2) + mean

gaussian_builtin = np.random.normal(mean, sigma, img.shape)

noisy_manual = np.clip(img.astype(float) + gaussian_manual, 0, 255).astype(np.uint8)

noisy_builtin = np.clip(img.astype(float) + gaussian_builtin, 0, 255).astype(np.uint8)

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(noisy_manual)
plt.title("Gaussian Noise - Equation")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(noisy_builtin)
plt.title("Gaussian Noise - Built-in")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.hist(gaussian_manual.ravel(), bins=50)
plt.title("Manual Gaussian Histogram")

plt.subplot(2, 2, 4)
plt.hist(gaussian_builtin.ravel(), bins=50)
plt.title("Built-in Gaussian Histogram")

plt.tight_layout()
plt.show()
```

```python
a = 0.5
b = 3

u = np.random.rand(*img.shape)

erlang_manual = np.zeros(img.shape)

for i in range(b):
    erlang_manual += -np.log(np.random.rand(*img.shape)) / a

erlang_builtin = np.random.gamma(b, 1/a, img.shape)

noisy_manual = np.clip(img.astype(float) + erlang_manual * 10, 0, 255).astype(np.uint8)

noisy_builtin = np.clip(img.astype(float) + erlang_builtin * 10, 0, 255).astype(np.uint8)

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(noisy_manual)
plt.title("Erlang Noise - Equation")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(noisy_builtin)
plt.title("Erlang Noise - Built-in")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.hist(erlang_manual.ravel(), bins=50)
plt.title("Manual Erlang Histogram")

plt.subplot(2, 2, 4)
plt.hist(erlang_builtin.ravel(), bins=50)
plt.title("Built-in Erlang Histogram")

plt.tight_layout()
plt.show()
```

```python
a = 0.5

u = np.random.rand(*img.shape)

exponential_manual = -np.log(1 - u) / a

exponential_builtin = np.random.exponential(1/a, img.shape)

noisy_manual = np.clip(img.astype(float) + exponential_manual * 10, 0, 255).astype(np.uint8)

noisy_builtin = np.clip(img.astype(float) + exponential_builtin * 10, 0, 255).astype(np.uint8)

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(noisy_manual)
plt.title("Exponential Noise - Equation")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(noisy_builtin)
plt.title("Exponential Noise - Built-in")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.hist(exponential_manual.ravel(), bins=50)
plt.title("Manual Exponential Histogram")

plt.subplot(2, 2, 4)
plt.hist(exponential_builtin.ravel(), bins=50)
plt.title("Built-in Exponential Histogram")

plt.tight_layout()
plt.show()
```

```python
a = 0
b = 200

u = np.random.rand(*img.shape)

rayleigh_manual = a + np.sqrt(-b * np.log(1 - u))

rayleigh_builtin = np.random.rayleigh(np.sqrt(b / 2), img.shape)

noisy_manual = np.clip(img.astype(float) + rayleigh_manual, 0, 255).astype(np.uint8)

noisy_builtin = np.clip(img.astype(float) + rayleigh_builtin, 0, 255).astype(np.uint8)

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(noisy_manual)
plt.title("Rayleigh Noise - Equation")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(noisy_builtin)
plt.title("Rayleigh Noise - Built-in")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.hist(rayleigh_manual.ravel(), bins=50)
plt.title("Manual Rayleigh Histogram")

plt.subplot(2, 2, 4)
plt.hist(rayleigh_builtin.ravel(), bins=50)
plt.title("Built-in Rayleigh Histogram")

plt.tight_layout()
plt.show()
```

## Conclusion

In this assignment, five types of noise distributions were generated: Uniform, Gaussian, Erlang, Exponential and Rayleigh.

Each noise was generated using mathematical equations and built-in NumPy functions. The noisy images and their histograms were plotted for comparison.

The results show that both methods produce approximately similar distributions. Histograms help us understand the shape and behaviour of different noise distributions.
